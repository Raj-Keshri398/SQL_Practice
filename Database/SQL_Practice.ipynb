```python
# Display employees name with depart name.

# tables - employees, dept_emp, departments

query = """
            Select
            e.emp_no,
            e.first_name,
            e.last_name,
            d.dept_name 
                    from employees e
            JOIN dept_emp de
                    ON e.emp_no = de.emp_no
            JOIN departments d
                    ON de.dept_no = d.dept_no
            limit 5;"""

df = pd.read_sql(query, engine)
df
```

```python
# Display employee name with salary

# tables - employees , salaries

query = """
SELECT
    e.first_name,
    e.last_name,
    s.salary
FROM employees e
JOIN salaries s
    ON e.emp_no = s.emp_no;
"""

df = pd.read_sql(query, engine)
df
```

```python
# Display employee name with titles

# tables - employees , titles

query = """
SELECT
    e.first_name,
    e.last_name,
    t.title
FROM employees e
JOIN titles t
    ON e.emp_no = t.emp_no;
"""

df = pd.read_sql(query, engine)
df
```

```python
# Display employee name, department name, and salary current date.

# tables -- employees, dept_emp, depatments, salaries

query = """
SELECT
    e.first_name,
    e.last_name,
    d.dept_name,
    s.salary
FROM employees e
JOIN dept_emp de
    ON e.emp_no = de.emp_no
JOIN departments d
    ON de.dept_no = d.dept_no
JOIN salaries s
    ON e.emp_no = s.emp_no
WHERE de.to_date = '9999-01-01'
AND s.to_date = '9999-01-01'
LIMIt 10;
"""

df = pd.read_sql(query, engine)
df
```

```python
# ================================= GROUP BY =============================
```

```python
# Count employees in each department

query = """ SELECT 
d.dept_name,
COUNT(e.emp_no) AS total_employees
FROM departments d
JOIN dept_emp de
ON d.dept_no = de.dept_no
JOIN employees e
ON de.emp_no = e.emp_no
GROUP BY d.dept_name;"""

df = pd.read_sql(query, engine)
df
```

```python
# Find theb Average Salary in each department.

query = """
SELECT 
d.dept_name,
ROUND(AVG(s.salary), 2) as average_salary
FROM departments d
JOIN dept_emp de
    ON d.dept_no = de.dept_no
JOIN employees e
    ON de.emp_no = e.emp_no
JOIN salaries s
    ON e.emp_no = s.emp_no
GROUP BY d.dept_name;"""

df = pd.read_sql(query, engine)
df
```

```python
# Find the Maxmium Salary of each department.

query = """
SELECT 
d.dept_name,
MAX(s.salary) as maximum_salary
FROM departments d
JOIN dept_emp de
    ON d.dept_no = de.dept_no
JOIN employees e
    ON de.emp_no = e.emp_no
JOIN salaries s
    ON e.emp_no = s.emp_no
GROUP BY d.dept_name;"""

df = pd.read_sql(query, engine)
df
```

```python
# Find the Minmum Salary of each department.

query = """
SELECT 
d.dept_name,
Min(s.salary) as minimum_salary
FROM departments d
JOIN dept_emp de
    ON d.dept_no = de.dept_no
JOIN employees e
    ON de.emp_no = e.emp_no
JOIN salaries s
    ON e.emp_no = s.emp_no
GROUP BY d.dept_name;"""

df = pd.read_sql(query, engine)
df
```

```python
# Display employees earning more than the average salary.

query = """
SELECT 
    e.emp_no, 
    e.first_name,
    e.last_name,
    s.salary
FROM employees e
JOIN salaries s
    ON e.emp_no = s.emp_no
WHERE s.salary > (SELECT AVG(s.salary) FROM salaries) limit 10;"""
df = pd.read_sql(query, engine)
df
```

```python
# Display employee name, deaprtment, current title, current salary

query = """
SELECT 
    e.emp_no,
    e.first_name,
    e.last_name,
    d.dept_name,
    t.title,
    s.salary
FROM employees e
JOIN dept_emp de
     ON e.emp_no = de.emp_no
JOIN departments d
     ON de.dept_no = d.dept_no
JOIN titles t
    ON e.emp_no = t.emp_no
JOIN salaries s
    ON e.emp_no = s.emp_no limit 10; """
df = pd.read_sql(query, engine)
df
```

```python
# Find top 5 highest paid employees.

query = """ SELECT 
    e.emp_no,
    e.first_name,
    e.last_name,
    MAX(s.salary) As salary
FROM employees e
JOIN salaries s
    ON e.emp_no = s.emp_no
ORDER BY s.salary DESC limit 5; """
df = pd.read_sql(query, engine)
df
```

```python
# Find the highest paid employee in each department

query = """
SELECT 
    e.emp_no,
    e.first_name,
    e.last_name,
    d.dept_name, 
    s.salary
FROM employees e
JOIN dept_emp de
    ON e.emp_no = de.emp_no
JOIN departments d
    ON de.dept_no = d.dept_no
JOIN salaries s
    ON e.emp_no = s.emp_no
WHERE s.salary = (
                SELECT MAX(s2.salary) FROM salaries s2
                JOIN dept_emp de2
                ON s2.emp_no = de2.emp_no
                 WHERE de2.dept_no = d.dept_no);"""


df = pd.read_sql(query, engine)
df
```
